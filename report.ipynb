# Choices and Reasoning

## Cleaning

For the cleaning we decided to remove impossible/near-impossible values. For missing values we decided to fill numerical values with the median of the other rows from the same column. Non-numerical values were filled using the mode. The mode and median are calculated after removing the impossible rows.
When working on the testing file, no rows will be dropped as the final submission needs all rows to work. This was the final method chosen as it proved to deliver the most reliable results we found when training.

Other methods tried were setting all negative values to 0 and only training on values in the 97th percentile to avoid outliers. We also tried removing all unclean data entirely and only using the clean existing data, but this removed too many rows.

# GenAI

Artificial intelligence was used to assist in this project. It was mainly used with help in writing code snippets. It was also used to get suggestions for different models other than random forest. The main thought process of the project and the choices made are our own choice and decision.

### AI Policy of the project

This policy does not forbid using GenAI tools. What it does require is that you remain the thinking agent behind your project.

GenAI can help you explore ideas, suggest alternatives, or get unstuck. But it cannot replace your understanding. Every choice you make (dataset, model, parameters, evaluation) must be your own decision, based on reasoning you can clearly explain.

If something appears in your report or code, you are expected to:

    know what it does
    understand why you used it
    be able to defend it when questioned

If you cannot explain or justify a decision, it means you did not truly make that decision yourself—and that is not acceptable for this course.

In short: Use GenAI as a tool, not as a crutch. Ownership and responsibility remain with you.
